```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("/content/Dataset for Data Analytics.xlsx")
```

```python
## first 5 rows
df.head()
```

```python
## basic information
df.info()
df.shape
df.dtypes
```

```python
## missing values
df.isnull().sum()
```

```python
## statistics
df.describe()
```

```python
## product chart
df["Product"].value_counts().plot(kind="bar")

plt.title("Products Sold")
plt.xlabel("Product")
plt.ylabel("Count")
plt.show()


## printer is the most ordered product and phone is the least one
```

```python
## order chart
df["OrderStatus"].value_counts().plot(kind="bar")

plt.title("Order Status")
plt.show()
```

```python
sns.boxplot(x=df["TotalPrice"])

plt.title("Outliers in Total Price")
plt.show()
# Detect outliers using a boxplot
```

```python
#  distribution of Total Price
sns.histplot(df["TotalPrice"], bins=20, kde=True)

plt.title("Distribution of Total Price")
plt.show()
```

```python
df["PaymentMethod"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Payment Methods")
plt.ylabel("")
plt.show()
## pie charrt
```

```python
corr = df[[
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]].corr()

corr
## correlation
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
## heatmap
```
