```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the Excel dataset
df = pd.read_excel("/content/Dataset for Data Analytics.xlsx")

# Display first 5 rows
df.head()
```

```python
# Count the number of orders for each order status
df["OrderStatus"].value_counts().plot(kind="bar")

# Add chart title and labels
plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

# Display the chart
plt.show()
```

```python
# Plot the distribution of payment methods
df["PaymentMethod"].value_counts().plot(kind="pie", autopct="%1.1f%%")

# Add chart title
plt.title("Payment Method Distribution")

# Remove default y-axis label
plt.ylabel("")

# Display the chart
plt.show()
```

```python
# Display the top 10 most purchased products
df["Product"].value_counts().head(10).plot(kind="bar")

# Add chart title and labels
plt.title("Top 10 Products Sold")
plt.xlabel("Product")
plt.ylabel("Number of Orders")

# Rotate product names for better readability
plt.xticks(rotation=45)

# Display the chart
plt.show()
```

```python
# Group data by month and calculate total sales
monthly_sales = df.groupby(df["Date"].dt.month)["TotalPrice"].sum()

# Plot monthly sales
monthly_sales.plot(kind="line", marker="o")

# Add chart title and labels
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

# Display the chart
plt.show()
```

```python
# Plot histogram of Total Price
df["TotalPrice"].plot(kind="hist", bins=20)

# Add chart title and labels
plt.title("Distribution of Total Price")
plt.xlabel("Total Price")
plt.ylabel("Frequency")

# Display the chart
plt.show()
```

```python
# Plot relationship between Quantity and Total Price
plt.scatter(df["Quantity"], df["TotalPrice"])

# Add chart title and labels
plt.title("Quantity vs Total Price")
plt.xlabel("Quantity")
plt.ylabel("Total Price")

# Display the chart
plt.show()
```

```python
# Count the number of orders for each cart size
df["ItemsInCart"].value_counts().sort_index().plot(kind="bar")

# Add chart title and labels
plt.title("Items in Cart Distribution")
plt.xlabel("Number of Items")
plt.ylabel("Number of Orders")

# Display the chart
plt.show()
```

```python
# Count the number of orders in each status
plt.figure(figsize=(6,4))
sns.countplot(x="OrderStatus", data=df)

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Count")

plt.show()
```

```python
# Plot distribution of Total Price
plt.figure(figsize=(7,4))
sns.histplot(df["TotalPrice"], bins=20, kde=True)

plt.title("Distribution of Total Price")
plt.xlabel("Total Price")
plt.ylabel("Frequency")

plt.show()
```

```python
# Box plot to identify outliers
plt.figure(figsize=(6,4))
sns.boxplot(y=df["TotalPrice"])

plt.title("Outliers in Total Price")

plt.show()
```

```python
# Relationship between Quantity and Total Price
plt.figure(figsize=(6,4))
sns.scatterplot(x="Quantity", y="TotalPrice", data=df)

plt.title("Quantity vs Total Price")

plt.show()
```

```python
# Find top 10 selling products
top_products = df["Product"].value_counts().head(10)

plt.figure(figsize=(9,5))
sns.barplot(x=top_products.index, y=top_products.values)

plt.title("Top 10 Products")
plt.xlabel("Product")
plt.ylabel("Orders")

plt.xticks(rotation=45)

plt.show()
```

```python
# ============================
# Final Conclusion
# ============================

# 1. The Order Status chart showed the distribution of different order statuses,
#    helping identify whether most orders were successfully delivered,
#    pending, or cancelled.

# 2. The Payment Method chart revealed the preferred payment methods used
#    by customers, making it easier to understand customer payment behavior.

# 3. The Top Products chart highlighted the products with the highest number
#    of purchases, indicating which items contributed the most to overall sales.

# 4. The Monthly Sales Trend chart showed how sales changed over time,
#    making it possible to identify periods of higher and lower sales activity.

# 5. The Total Price distribution illustrated that most orders were concentrated
#    within a specific price range, while only a few orders had exceptionally
#    high or low values.

# 6. The Scatter Plot between Quantity and Total Price indicated a positive
#    relationship, suggesting that orders with higher quantities generally
#    resulted in higher total prices.

# Overall, the visualizations provided meaningful insights into customer
# purchasing behavior, sales performance, payment preferences, and product
# demand, making the dataset easier to analyze and interpret.
```
